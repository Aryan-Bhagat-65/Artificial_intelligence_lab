```python
import math
import random


class BoardState:

    def __init__(self, state=None):
        # A state is represented as a list of 8 integers.
        # Index = column, Value = row position of the queen.
        if state is None:
            self.state = [random.randint(0, 7) for _ in range(8)]
        else:
            self.state = list(state)
        self.cost = self.calculate_cost()

    def calculate_cost(self):
        """Calculates the number of attacking pairs of queens (conflicts).

        Lower cost is better (0 means solution found).
        """
        conflicts = 0
        for i in range(8):
            for j in range(i + 1, 8):
                # Same row conflict
                if self.state[i] == self.state[j]:
                    conflicts += 1
                # Diagonal conflict
                elif abs(self.state[i] - self.state[j]) == abs(i - j):
                    conflicts += 1
        return conflicts

    def get_random_neighbor(self):
        """Generates a neighbor by moving one random queen to a random new

        row.
        """
        neighbor_state = list(self.state)
        col = random.randint(0, 7)
        row = random.randint(0, 7)

        # Ensure we actually change the row
        while row == neighbor_state[col]:
            row = random.randint(0, 7)

        neighbor_state[col] = row
        return BoardState(neighbor_state)


def simulated_annealing():
    # 1: current <- initial state
    current = BoardState()

    # 2: T <- a large positive value
    T = 100.0
    cooling_rate = 0.995  # Factor to decrease T

    # 3: while T > 0 do (using a small epsilon threshold near 0)
    while T > 0.001:
        if current.cost == 0:
            break  # Found a perfect solution early

        # 4: next <- a random neighbour of current
        next_state = current.get_random_neighbor()

        # 5: ΔE <- current.cost - next.cost
        delta_E = current.cost - next_state.cost

        # 6: if ΔE > 0 then
        if delta_E > 0:
            # 7: current <- next
            current = next_state
        # 8: else
        else:
            # 9: current <- next with probability p = e^(ΔE / T)
            p = math.exp(delta_E / T)
            if random.random() < p:
                current = next_state

        # 11: decrease T
        T *= cooling_rate

    # 13: return current
    return current


# --- Execution and Visualization ---
if __name__ == "__main__":
    solution = simulated_annealing()

    print("--- 8-Queens Simulated Annealing Result ---")
    print(f"Final State (Rows per column): {solution.state}")
    print(f"Remaining Conflicts (Cost): {solution.cost}\n")

    print("Visual Board Layout:")
    for row in range(8):
        line = ""
        for col in range(8):
            if solution.state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
```
